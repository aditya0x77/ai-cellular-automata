```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd

board = np.random.randint(0, 2, (10, 10))
```

```python
def count_neighbors(board, row, col):
    neighbors = 0

    if(0 <= row - 1 < board.shape[0] and 0 <= col - 1 < board.shape[1]):
        if(board[row-1,col-1] == 1):
             neighbors += 1

    if(0 <= row - 1 < board.shape[0] and 0 <= col < board.shape[1]):
        if(board[row-1,col] == 1):
             neighbors += 1

    if(0 <= row - 1 < board.shape[0] and 0 <= col + 1 < board.shape[1]):
        if(board[row-1,col+1] == 1):
             neighbors += 1

    if(0 <= row  < board.shape[0] and 0 <= col - 1 < board.shape[1]):
        if(board[row,col-1] == 1):
             neighbors += 1

    if(0 <= row < board.shape[0] and 0 <= col +1 < board.shape[1]):
        if(board[row,col+1] == 1):
             neighbors += 1

    if(0 <= row + 1 < board.shape[0] and 0 <= col - 1 < board.shape[1]):
        if(board[row+1,col-1] == 1):
             neighbors += 1

    if(0 <= row + 1 < board.shape[0] and 0 <= col < board.shape[1]):
        if(board[row+1,col] == 1):
             neighbors += 1

    if(0 <= row + 1 < board.shape[0] and 0 <= col + 1 < board.shape[1]):
        if(board[row+1,col+1] == 1):
             neighbors += 1


    return neighbors
    
    
```

```python
def generate_next(board):
    next_board = board.copy()
    for i in range(board.shape[0]):
        for j in range(board.shape[1]):
            neighbours = count_neighbors(board,i,j)
            if(neighbours < 2 and board[i,j] == 1):
                next_board[i,j] = 0
            elif(2 <= neighbours <=3 and board[i,j] == 1):
                next_board[i,j] = 1
            elif(neighbours > 3 and board[i,j] == 1):
                next_board[i,j] = 0
            elif(neighbours == 3 and board[i,j] == 0):
                next_board[i,j] = 1
    return next_board

            
            
```

```python
X = []
Y = []
```

```python
for i in range(1000):
    board = np.random.randint(0,2, (32, 32))
    for j in range(100):
        next = generate_next(board)
        X.append(board)
        Y.append(next)
        board = next
```
